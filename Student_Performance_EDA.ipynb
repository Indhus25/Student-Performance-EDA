```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

print("Libraries imported successfully!")
```

```python
df = pd.read_csv('/content/StudentPerformanceFactors.csv')

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Data Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()

print("\nMemory Usage:")
print(df.memory_usage(deep=True))

print("\nFirst 5 Rows:")
display(df.head())

print("\nLast 5 Rows:")
display(df.tail())
```

```python
missing = df.isnull().sum()

print(missing[missing > 0])

plt.figure(figsize=(12,5))
missing[missing > 0].plot(kind='bar')

plt.title("Missing Values by Column")
plt.xlabel("Columns")
plt.ylabel("Missing Values")

plt.show()
```

```python
df.describe()
```

```python
df['Gender'].value_counts()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['Exam_Score'], bins=20, edgecolor='black')

plt.title("Distribution of Exam Scores")
plt.xlabel("Exam Score")
plt.ylabel("Number of Students")

plt.show()
```

The histogram shows the distribution of students' exam scores. It helps identify the most common score range and understand how student performance is distributed.

```python
plt.figure(figsize=(8,5))

plt.scatter(df['Hours_Studied'], df['Exam_Score'], alpha=0.5)

plt.title("Study Hours vs Exam Score")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")

plt.show()
```

The scatter plot shows the relationship between study hours and exam scores. It helps identify whether students who study more tend to achieve higher scores.

```python
plt.figure(figsize=(8,5))

sns.boxplot(x='Gender', y='Exam_Score', data=df)

plt.title("Exam Score Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Exam Score")

plt.show()
```

The boxplot compares exam score distributions across genders. It helps identify differences in median scores, score variability, and potential outliers.

```python
plt.figure(figsize=(8,5))

sns.boxplot(x='Parental_Involvement', y='Exam_Score', data=df)

plt.title("Parental Involvement vs Exam Score")
plt.xlabel("Parental Involvement")
plt.ylabel("Exam Score")

plt.show()
```

This visualization compares exam scores across parental involvement levels. It helps explore whether students with greater parental involvement tend to achieve different scores.

```python
plt.figure(figsize=(8,5))

sns.scatterplot(x='Sleep_Hours', y='Exam_Score', data=df)

plt.title("Sleep Hours vs Exam Score")
plt.xlabel("Sleep Hours")
plt.ylabel("Exam Score")

plt.show()
```

The scatter plot explores the relationship between sleep duration and exam scores. It helps identify whether sleep hours are associated with student performance.

```python
print("KEY FINDINGS")

print("1. Average Exam Score:",
      round(df['Exam_Score'].mean(), 2))

print("2. Median Exam Score:",
      df['Exam_Score'].median())

print("3. Highest Exam Score:",
      df['Exam_Score'].max())

print("4. Lowest Exam Score:",
      df['Exam_Score'].min())

print("5. Study Hours Correlation:",
      round(df['Hours_Studied'].corr(df['Exam_Score']), 2))
```

## Key Findings

1. The average exam score of students is 67.24, while the median score is 67.
2. The highest recorded exam score is 101, and the lowest is 55.
3. The correlation between study hours and exam scores is 0.45, indicating a moderate positive relationship.
4. Students who spend more time studying generally tend to achieve higher exam scores.
5. The average and median scores are very close, suggesting a relatively balanced score distribution.
